```python
#3. Check whether 45 is greater than 30 and less than 50.
a=45
b=30
c=50
if a>b and a<c:
    print("45 is greater than 30 and less than 50")
else:
    print("45 is not greater than 30 and not less than 50")
```

```python
#4. Check whether the price ₹850 is less than or equal to ₹1000.
p1=850
p2=1000
if p1<=p2:
    print("₹850 is less ₹1000")
else:
    print("₹850 is not less than or equal to ₹1000")
```
