```python
#calculate the electricity bill based on units consumed using slabs.
units = int(input("Enter the number of units consumed: "))
#First 100 units → ₹1.50 per unit
#Next 100 units (101–200) → ₹2.50 per unit
#Next 100 units (201–300) → ₹4.00 per unit
#Above 300 units → ₹6.00 per unit
if units <=100:
    print(1.50*units)
elif units >=101 and units<=200:
    print(2.50*units)
elif units >=201 and units<=300:
    print(4*units)
elif units >300:
    print(6*units)
else:
    print("invalid")
```

```python
#Calculate income tax based on salary slabs.

#Tax Slabs
#Up to ₹2,50,000 → No tax
#₹2,50,001 to ₹5,00,000 → 5%
#₹5,00,001 to ₹10,00,000 → 20%
#Above ₹10,00,000 → 30%

salary=int(input("enter annual salary:"))
if salary <=250000:
    print("tax=0")
elif salary >=250001 and salary <=500000:
    print(salary*5/100)
elif salary >=500001 and salary <=1000000:
    print(salary*20/100)
else:
    print(salary*30/100)
```
