# ASSIGNMENT OPERATOR


```python
#assign a value to a variable and increase it by 10 using an assignment operator
x=int(input("enter a value:"))
x+=10
print(x)
```

```python
#store a number and decrese it by 15 using an assignment operator
x=int(input("enter a value:"))
x-=15
print(x)
```

```python
#multiply a number by 5 using an assignment operator
x=int(input("enter a value:"))
x*=5
print(x)
```

```python
#divide a number by 4 using an assignment operator
x=int(input("enter a value:"))
x/=4
print(x)
```

```python
#find the reminder when a number is divided by 7 using an assignment operator
x=int(input("enter a value:"))
x%=7
print(x)
```

```python
#raise a number to the power of 3 using an assignment operator
x=int(input("enter a value:"))
x**=3
print(x)
```

```python
#perform floor division of a number by 5 using an assignment operator
x=int(input("enter a value:"))
x//=5
print(x)
```

```python
#swap the values of two variables without using a thir variable
x=int(input("enter a value:"))
y=int(input("enter another value:"))
x,y=y,x
print("x:",x)
print("y:",y)
```

```python
#add the marks of three subjects one by one using assignment operator
a=int(input("subject 1:"))
b=int(input("subject 2:"))
c=int(input("subject 3:"))
print("total marks:",a+b+c)

#another method
total=85
total+=90
total+=97
print("total mark:",total)
```

```python
#update the balance after depositing and withdrawing money using assignment operators
x=int(input("current balance:"))

#deposite
y=int(input("deposite:"))
x+=y
print("after deposite:",x)

#withdraw
a=int(input("withdraw:"))
x-=a
print("after withdraw:",x)
```
