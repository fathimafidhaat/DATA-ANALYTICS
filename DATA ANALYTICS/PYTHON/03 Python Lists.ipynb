# PYTHON LISTS

```python
#CREATE A LIST
list=["apple","banana","cherry"]
print(list)

#add 2nd item "pineapple" to the list
list.insert(1,"pineapple")
print(list)

#remove "pineapple" from the list
list.remove("pineapple")
print(list)

#replace "banana" to "kiwi"
list[1]="kiwi"
print(list)

#remove the 2nd item in the list
list.pop(1)
print(list)
```

```python
list=[10,50,32,12,47,63,90]

#no.of items    
x=(len(list))
print("no of items:",x)

#total value
y=sum(list)
print("total:",y)

#highest value
a=max(list)
print("highest value is:",a)

#lowest value
b=min(list)
print("lowest value is:",b)

#average
c=sum(list)/(len(list))
print("average value is:",c)
```

```python
#sort the list into ascending order
list=[10,50,32,12,47,63,90]
list.sort()
print("ascending order:",list)

#sort to descending order
list.sort(reverse=True)
print("descending order:",list)
```

```python
#reverse the list
list=[10,50,32,12,47,63,90]
list.reverse()
print(list)
```

```python
#print second largest number
list=[10,50,32,12,47,63,90]
x=max(list)
print("largest is:",x)
list.remove(x)
y=max(list)
print("second largest is:",y)

#print 2nd largset number(another method)
list=[10,50,32,12,47,63,90]
list.sort()
print(list)
print("second largest is:",list[-2])
      
#print 2nd largset number(another method)
list=[10,50,32,12,47,63,90]
list.sort(reverse=True)
print(list)
print("2nd largest :",list[1])
```

```python
#print second smallest number
list=[10,50,32,12,47,63,90]
a=min(list)
print("smallest :",a)
list.remove(a)
b=min(list)
print("second smallest is:",b)

#print 2nd smallest number(another method)
list=[10,50,32,12,47,63,90]
list.sort()
print(list)
print("2nd smallest is:",list[1])

#print 2nd smallest number(another method)
list=[10,50,32,12,47,63,90]
list.sort(reverse=True)
print(list)
print("second smallest is:",list[-2])
```
