### STRING SLICING


```python
#1. Print the first five characters of the string "Artificial Intelligence" .
txt="Artificial Intelligence"
print(txt[:5])
```

```python
#2. Print the last four characters of the string "Technology"  .
x="Technology"
print(x[6:])
```

```python
#3. Print the characters from index 2 to index 7 of the string "Programming" .
x="Programming"
print(x[2:7])
```

### MODIFY STRINGS

```python
#1. Convert the string "python programming" to uppercase.
x="python programming"
print(x.upper())
```

```python
#2. Convert the string "DATA SCIENCE" to lowercase.
x="DATA SCIENCE"
print(x.lower())
```

```python
#3. Remove the leading and trailing spaces from the string " Hello Python " .
x=" Hello Python "
print(x.strip())
```

```python
#4. Replace "Python" with "Java" in the string "I am learning Python" .
x="I am learning Python"
print(x.replace("Python","Java"))
```

```python
#5. Split the string "Python is easy to learn" into a list of words.
x="Python is easy to learn"
print(x.split())
```
