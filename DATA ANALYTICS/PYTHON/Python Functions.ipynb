# PYTHON FUNCTION

```python
#python function by using print
def add(a,b):
    print(a+b)
add(10,20)
```

```python
#python function by using return
def add(a,b):
    return a+b
result=add(10,20)
print(result)
print(result+result)
print(result**2)
print(result+70)
print(result**3)
```
