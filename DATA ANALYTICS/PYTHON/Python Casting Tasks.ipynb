```python
#1. Take two numbers as input (as strings), convert them to integers, and display their sum, difference, product, and quotient.
num1=input("enter a number:")
num2=input("enter a number:")
x=int(num1)
y=int(num2)
print(x)
print(y)

z=x+y
d=x-y
p=x*y
q=x/y
print("sum:",z)
print("difference:",d)
print("product:",p)
print("quotient:",q)
```

```python
#2. Convert the string "123.75" into a float, then convert it to an integer, and display both values.
st="123.75"
x=float(st)
y=int(x)
print("float:",x)
print("integer:",y)
```

```python
#3. Convert the integer 250 into a float, string, boolean, and complex number.Display the value and data type after each conversion.
intt=250
x=float(intt)
y=str(intt)
z=bool(intt)
a=complex(intt)
print(x,"float")
print(y,"string")
print(z,"boolean")
print(a,"complex")
```
