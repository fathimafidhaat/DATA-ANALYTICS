```python
#create a set of students in football and cricket team
football={"vinu","anu","ajay","roshan","amal"}
cricket={"anu","vishal","amal","jay","manu"}

#football players
print("football players:",football)

#cricket players
print("cricket players:",cricket)

#show only play football
onlyfootball=football.difference(cricket)
print("only play football:",onlyfootball)

#show only play cricket
onlycricket=cricket.difference(football)
print("only play cricket:",onlycricket)

#show both items playing students
include2=football&cricket
print("including 2 items:",include2)
```
