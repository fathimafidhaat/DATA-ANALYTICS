```python
#15.Count the number of uppercase and lowercase letters in "Python Data Science"
s="Python Data Science"
upper=0
lower=0
for char in s:
    if char.isupper():
        upper+=1
    elif char.islower():
        lower+=1
print("upper:",upper)
print("lower:",lower)
```
