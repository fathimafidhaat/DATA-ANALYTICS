# ARRAY SLICING

```python
a=np.array([10,20,30,40,50])
print(a[1:4])
```

# ARRAY ITERATING

```python
a=np.array([1,2,3,4,5])
for i in a:
    print(i)
```

```python
a=np.array([[10,20,30],
           [40,50,60]])
for i in np.nditer(a):
    print(i)
```

# JOINING ARRAYS

```python
a=([1,2,3])
b=([4,5,6])
print(np.concatenate((a,b)))
```

```python
#another methods for joining
#stack()_joining arrays vertically
#vstack()_joining arrays vertically
#hstack()_joing arrays horizontaly
```

### 1.stack()

```python
a=([1,2,3])
b=([4,5,6])
print(np.stack((a,b)))
```

### 2.vstack()

```python
a=([1,2,3])
b=([4,5,6])
print(np.vstack((a,b)))
```

### 3.hstack()

```python
a=([1,2,3])
b=([4,5,6])
print(np.hstack((a,b)))
```

# SPLITING OF ARRAYS

```python
a=np.array([1,2,3,4,5])
print(np.array_split(a,2))
```

# SEARCHING ARRAYS

```python
a=np.array([10,20,30,40])
print(np.where(a==20))
```

```python
a=np.array([10,20,30,40])
print(np.where(a==30))
```

```python
a=np.array([10,20,30,20,40])
print(np.where(a==20))
```

```python
a=np.array([10,20,30,40])
print(np.where(a>20))
```

# SORTING ARRAYS

```python
a=np.array([40,60,30,20,10])
print(np.sort(a))
```

```python
a=np.array(["banana","cherry","apple"])
print(np.sort(a))
```

# FILTER ARRAYS

```python
a=np.array([40,60,30,20,10])
print(a[a>25])
```

```python
a = np.array([1, 2, 3, 4, 5, 6])
print(a[a%2==0])
```

```python
a = np.array([1, 2, 3, 4, 5, 6])
print(a[a%2!=0])
```

# ARRAY MATHEMATICS

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**b)
```

```python
print(a*2)
print(a/2)
print(a**2)
```

# AXIS CONCEPT

```python
#axis=0(column)
#axis=1(row)
a = np.array([[1, 2, 3],
              [4, 5, 6]])
print(np.sum(a,axis=0))
print(np.sum(a,axis=1))
```
