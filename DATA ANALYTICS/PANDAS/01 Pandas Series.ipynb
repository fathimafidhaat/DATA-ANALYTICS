# SORTING A SERIES

### 1.Asceding Order

```python
a=pd.Series([20,15,35,50,40])
print(a.sort_values())
```

### 2.Descending Order

```python
a=pd.Series([20,15,35,50,40])
print(a.sort_values(ascending=False))
```

## Resetting the Index After Sorting


```python
a=pd.Series([20,15,35,50,40])
result=a.sort_values().reset_index(drop=True)
print(result)
```

# MISSING VALUES IN A SERIES

```python
import pandas as pd
import numpy as np

s=pd.Series([10,20,30,np.nan,50,np.nan])
print(s)
```

# CHECKING FOR MISSING VALUES-isna()

```python
print(s.isna())
```

# REMOVING MISSING VALUES-dropna()

```python
print(s.dropna())
```

# FILLING MISSING VALUES-fillna()

```python
s=pd.Series([10,20,30,np.nan,50,np.nan])
x=s.fillna(0)
print(x)
```

# ADDING TWO SERIES TOGETHER
