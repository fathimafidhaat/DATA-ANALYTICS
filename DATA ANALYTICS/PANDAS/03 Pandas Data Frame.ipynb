# PANDAS DATAFRAME

```python
import pandas as pd
```

## DATA FRAME FROM DICTIONARY

```python
data={
    "Name":["fidha","shibna","yusra"],
    "Age":[21,22,20],
    "Mark":[90,88,95]
}
df=pd.DataFrame(data)
print(df)
```

## DATA FRAME FROM LIST

```python
data=[
    ["fidha",21,90],
    ["shibna",22,88],
    ["yusra",20,95]
]
df=pd.DataFrame(data,columns=["Name","Age","Mark"])
print(df)
    
```

## DATA FRAME WITH CUSTOM INDEX


```python
df=pd.DataFrame(data,index=["s1","s2","s3"])
print(df)
```

## DISPLYING COLUMNS


```python
print(df.columns)
```

## SELECTING A SINGLE COLUMN

```python
data={
    "Name":["fidha","shibna","yusra"],
    "Age":[21,22,20],
    "Mark":[90,88,95]
}
df=pd.DataFrame(data)
print(df["Name"])
```

## SELECTING MULTIPLE COLUMNS

```python
data={
    "Name":["fidha","shibna","yusra"],
    "Age":[21,22,20],
    "Mark":[90,88,95]
}
df=pd.DataFrame(data)
print(df[["Name","Age"]])
```

## SELECTIING ROWS USING iloc()

```python
df=pd.DataFrame(data)
print(df.iloc[0])
```

## SELECTING MULTIPLE ROWS

```python
df=pd.DataFrame(data)
print(df.iloc[:2])
```

## SELECTING A PARTICULAR ROW AND COLUMN

```python
df=pd.DataFrame(data)
print(df.iloc[1,2])
```

## CHECKING FOR MISSING VALUES-isna()

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
 "Name": ["Anu", "Rahul", "Priya"],
 "Mark": [85, np.nan, 90]
})
print(df.isna())
```

## COUNTING MISSING VALUES

```python
df.isna().sum()
```

## FILLING MISSING VALUES-fillna()

```python
df["Mark"]=df["Mark"].fillna(0)
print(df)
```

### REMOVING ROWS WITH MISSING VALUES-dropna()

```python
df=df.dropna()
print(df)
```

## VIEWING FIRST AND LAST ROWS-head(),tail()

```python
print(df.head()) 
print(df.head(3)) 
print(df.tail()) 
print(df.tail(2))
```
