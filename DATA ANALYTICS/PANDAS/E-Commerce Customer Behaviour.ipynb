# <center>FILL MISSING VALUES

```python
#fill missing age with mean age
df["age"].fillna(df["age"].mean(),inplace=True)
```

### Observation:
Missing age values are replaced with the mean age.

```python
df["traffic_source"].fillna("Not Available",inplace=True)
```

### Observation:
Missing traffic-source values are replaced with “Not Available”.

```python
df["session_duration_minutes"].fillna(df["session_duration_minutes"].mean(),inplace=True)
```

### Observation:
Missing session-duration values are replaced with the mean session duration.

```python
df["pages_viewed"].fillna(df["pages_viewed"].mean(),inplace=True)
```

### Observation:
Missing pages-viewed values are replaced with the mean number of pages viewed.

```python
df["product_price"].fillna(df["product_price"].mean(),inplace=True)
```

### Observation:
Missing product-price values are replaced with the mean product price.

```python
df["payment_method"].fillna("Not Available",inplace=True)
```

### Observation:
Missing payment-method values are replaced with “Not Available”.

```python
#recheking if any null values exist
df.isnull().sum()
```

### Observation:
After treatment, all columns have zero missing values

```python
#check for count of duplicate values
df.duplicated().sum()
```

### Observation:
The dataset contains 0 duplicate records.
