## MERGING DATA USING PANDAS

### BASIC MERGING

```python
import pandas as pd
```

```python
df1=pd.DataFrame({
    "id":[1,2,3],
    "name":["asha","binu","cathy"]
})

df2=pd.DataFrame({
    "id":[1,2,4],
    "marks":[90,80,85]
})

merged_df=pd.merge(df1,df2,on="id")
print(merged_df)
```

## TYPES OF JOINS

### 1.Inner Join(default)
#### Only Common Raws:

```python
pd.merge(df1,df2,on="id",how="inner")
```

### 2.Left Join
#### All Raws From Left DataFrame:

```python
pd.merge(df1,df2,on="id",how="left")
```

### 3.Right Join
#### All Raws From Right DataFrame:

```python
pd.merge(df1,df2,on="id",how="right")
```

### 4.Outer Join
#### All Rows From Both:

```python
pd.merge(df1,df2,on="id",how="outer")
```

## Merge On Different Column Names

```python
df1=pd.DataFrame({
    "id":[1,2,3],
    "name":["asha","binu","cathy"]
})

df2=pd.DataFrame({
    "student_id":[1,2,4],
    "marks":[90,80,85]
})
merged_df=pd.merge(
    df1,
    df2,
    left_on="id",
    right_on="student_id"
)
merged_df
```

## CONCATENATION
### Row Wise:

```python
pd.concat([df1,df2],axis=0)
```

### Column Wise:


```python
pd.concat([df1,df2],axis=1)
```

# APPLY

```python
marks=pd.Series([45,60,75,90])
def grade(x):
    if x>50:
        return"pass"
    else:
        return"fail"
print(marks.apply(grade))
```

#### Apply Using Lambda Function

```python
marks=pd.Series([45,60,75,90])
result=marks.apply(lambda x:x+5)
print(result)
```

### Apply On DataFrame Columns

```python
df=pd.DataFrame({
    "math":[80,90,95],
    "science":[89,98,70]
})
print(df.apply(max))
```
