### Drop rows with missing values df.dropna()


```python
#fill missing age with mean age
df["Age"].fillna(df["Age"].mean(),inplace=True)
```

```python
#fill missing embarked with mode of embarked
df["Embarked"].fillna(df["Embarked"].mode(0), inplace=True)
```

```python
df["Cabin"].fillna("Unknow",inplace=True)
```

```python
#rename column
df.rename(columns={"Sex":"Gender"})
```

```python
#add new column age+5
df["Age_after_5_years"]=df["Age"]+5
```
