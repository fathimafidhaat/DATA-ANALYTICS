## Handle Missing Values

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)
df["Cabin"].fillna("Unknown", inplace=True)

df.isnull().sum()
```

## Remove duplicates

```python
print(df.duplicated().sum())
```

## Feature Selection

```python
df.drop(columns=["PassengerId","Name","Ticket"], inplace=True)
df.head()
```

## Outlier Detection

```python
columns = df.select_dtypes(include = "number").columns

# Calculate Z-score
z_scores = np.abs(zscore(df[columns]))

# Identify outliers
outlier_counts = (z_scores > 3).sum(axis=0)

# Print column name and number of outliers
for i in range(len(columns)):
 print(columns[i], ":", outlier_counts[i], "outliers")
```

## The actual outlier values

```python
for column in columns:

 z = np.abs(zscore(df[column]))

 print("\n", column)
 print(df[column][z > 3])
```

## Outlier Handling

```python
outlier_cols = ["Age","Fare"]
```

```python
df_clean = df.copy()
for column in outlier_cols:

 z = np.abs(zscore(df_clean[column]))

 median = df_clean[column].median()

 df_clean.loc[z > 3, column] = median
```

```python
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```

## Outlier Removal

```python
df_clean = df.copy()
for column in outlier_cols:
 z = np.abs(zscore(df_clean[column]))
 df_clean = df_clean[z <= 3]
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```
