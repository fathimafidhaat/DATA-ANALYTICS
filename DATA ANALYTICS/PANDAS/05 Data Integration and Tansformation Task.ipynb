```python
import pandas as pd
```

## Basic Merging:

```python
df1=pd.DataFrame({
    "ID":[1,2,3,4,5],
    "Class":[10,9,10,8,9],
    "Name":["fidha","jalva","mufeeda","shibna","yusra"]
})
df2=pd.DataFrame({
    "ID":[1,2,3,6,7],
    "Marks":[90,96,80,78,95],
    "Grade":["A+","A+","A","B","A+"]
})
pd.merge(df1,df2,on="ID")
```

## Inner Join:


```python
pd.merge(df1,df2,on="ID",how="inner")
```

## Left Join:

```python
pd.merge(df1,df2,on="ID",how="left")
```

## Right Join:

```python
pd.merge(df1,df2,on="ID",how="right")
```

## Outer Join:

```python
pd.merge(df1,df2,on="ID",how="outer")
```

## Merge On Different Column Names

```python
df1=pd.DataFrame({
    "ID":[1,2,3,4,5],
    "Class":[10,9,10,8,9],
    "Name":["fidha","jalva","mufeeda","shibna","yusra"]
})
df2=pd.DataFrame({
    "Student_ID":[1,2,3,6,7],
    "Marks":[90,96,80,78,95],
    "Grade":["A+","A+","A","B","A+"]
})
pd.merge(df1,
         df2,
         left_on="ID",
         right_on="Student_ID"
)
```

## CONCATENATION

```python
pd.concat([df1,df2],axis=0)
```

```python
pd.concat([df1,df2],axis=1)
```

## APPLY


```python
mark=pd.Series([80,89,76,82,90])

def grade(x):
    if x>50:
        return"pass"
    else:
        return"fail"
print(mark.apply(grade))
```

### Apply Using Lambda Function

```python
mark=pd.Series([80,89,76,82,90])
result=mark.apply(lambda x:x+5)
print(result)
```

### Apply On DataFrame Columns

```python
df=pd.DataFrame({
    "maths":[60,90,70,69,80],
    "english":[90,80,78,88,95]
})
print(df.apply(max))
```
