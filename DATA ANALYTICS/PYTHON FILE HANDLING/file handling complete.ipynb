# CREATE AND WRITE TO FILE

```python
with open("example.txt","w")as file:
  file.write("welcome to the file")
```

# READ FILE

```python
with open("example.txt","r")as file:
    print(file.read())
```

# OVERWRITE FILE

```python
with open("example.txt","w")as file:
    file.write("hello!! this is python")
```

# APPEND NEW LINE

```python
with open("example.txt","a")as file:
    file.write("\nwelcome to the file")
```

# READ FULL CONTENT AGAIN

```python
with open("example.txt","r")as file:
    print(file.read())
```

# READ LINE BY LINE

```python
with open("example.txt","r")as file:
    for line in file:
        print(line.strip())
```

# READ ONLY ONE LINE 

```python
with open("example.txt","r")as file:
    print(file.readline())
```

# WORKING WITH FILE POINTER

```python
with open("example.txt","r")as file:
 print(file.read(5))
 file.seek(0)
 print(file.read())
```

# TELL CURRENT POSITION

```python
with open("example.txt","r")as file:
    print(file.tell())
    file.read(10)
    print(file.tell())
```

# READ LINES AS LIST

```python
with open("example.txt","r")as file:
    lines=file.readlines()
    print(lines)
```

# WRITE MULTIPLE LINES

```python
lines=["fidha\n","jalva\n ","mufeeda\n"]

with open("example.txt","a")as file:
    file.writelines(lines)
```

# COPY FILE

```python
with open("example.txt","r")as src:
    content=src.read()
    with open("copy.txt","w")as dest:
        dest.write(content)
```

# RENAME FILE

```python
import os
os.rename("example.txt","new_example.txt")
```

# DELETE FILE

```python
import os
os.remove("copy.txt")
```

# WORKING WITH CSV FILES

```python
import csv

with open("data.csv","w",newline="")as file:
    text=csv.writer(file)
    text.writerow(["name","age"])
    text.writerow(["fidha",21])
```

# WORKING WITH JSON

```python
import json
data={"name":"fidha","age":21}
with open("data.json","w")as file:
    json.dump(data,file)
    
with open("data.json","r")as file:
    print(json.load(file)) 
```
